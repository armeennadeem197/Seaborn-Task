## Task 1

Create this DataFrame:

```
Department    Salary
IT            70000
HR            50000
IT            85000
Sales         55000
Sales         60000
HR            52000
```

Create a `barplot()` showing salary by department.



````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data ={
    "Department": ["HR", "IT", "Finance", "HR", "IT", "Finance"],
    "Salary": [50000, 60000, 55000, 70000, 65000, 75000]
}

df = pd.DataFrame(data)
sns.barplot(data=df,x="Department" , y= "Salary")
plt.title("Data")

plt.show()
````

## Task 2

Using the same data, create a `countplot()` for departments.

````python
sns.countplot(data=df, x="Department")

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
````

## Task 3

Create a scatter plot using:

```
Experience = 1, 2, 3, 4, 5, 6
Salary     = 30, 40, 50, 65, 75, 90
```

````python
data = {
    "Experience": [1, 2, 3, 4, 5, 6],
    "Salary": [30, 40, 50, 65, 75, 90]
}

df = pd.DataFrame(data)

sns.scatterplot(data=df, x="Experience", y="Salary")

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
````

## Task 4

Create a histogram for:

```
45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90, 95
```

Use:

```python
kde=True
```

````python
marks = [45, 50, 55, 60, 60, 65, 70, 75, 80, 85, 90, 95]

df = pd.DataFrame({
    "Marks": marks
})

sns.histplot(data=df, x="Marks", kde=True)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Count")

plt.show()
````

## Task 5

Create a DataFrame containing:

```
Study_Hours
Attendance
Marks
```

Calculate correlation and display it using a heatmap.

````python
data = {
    "Study_Hours": [2, 3, 4, 5, 6, 7],
    "Attendance": [60, 65, 70, 80, 90, 95],
    "Marks": [50, 55, 65, 75, 85, 95]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(
    correlation,
    annot=True
)

plt.title("Correlation Heatmap")

plt.show()
````
